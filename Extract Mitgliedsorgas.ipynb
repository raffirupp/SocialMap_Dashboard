```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
import time

print("✅ Selenium & BeautifulSoup importiert.")
```

```python
# Chrome im Headless-Modus starten (läuft ohne sichtbares Fenster)
options = Options()
options.add_argument("--headless")          
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

# ChromeDriver automatisch herunterladen und starten
driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=options
)

url = "https://www.paritaet-berlin.de/mitglieder/unsere-mitgliedsorganisationen"
driver.get(url)

# etwas warten, bis JavaScript alles geladen hat
time.sleep(5)

# kompletten HTML-Inhalt der Seite holen
html = driver.page_source
driver.quit()

print("✅ Seite geladen – HTML-Länge:", len(html))
```

```python
soup = BeautifulSoup(html, "html.parser")

# Suche alle Einträge mit class="member-list-item"
entries = soup.select("div.member-list-item")

print(f"Gefundene Organisationseinträge: {len(entries)}")

# Zeige die ersten 5 als Vorschau
for e in entries[:5]:
    name = e.select_one("h2")
    link = e.select_one("a")
    print("→", name.get_text(strip=True) if name else None)
    print("   ", "https://www.paritaet-berlin.de" + link["href"] if link else None)
```

```python
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

options = Options()
options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=options
)

url = "https://www.paritaet-berlin.de/mitglieder/unsere-mitgliedsorganisationen"
driver.get(url)

# ⏳ Warte bis mindestens ein Element mit class="member-list-item" vorhanden ist
try:
    WebDriverWait(driver, 15).until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "div.member-list-item"))
    )
    print("✅ Organisationseinträge gefunden (Seite fertig geladen)")
except:
    print("⚠️ Timeout – keine Organisationen nach 15s gefunden")

html = driver.page_source
driver.quit()
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from time import sleep

BASE_URL = "https://www.paritaet-berlin.de/mitglieder/unsere-mitgliedsorganisationen"
results = []

# Begriffe, die wir NICHT im Ergebnis haben wollen
blacklist = [
    "Hier finden Sie eine Übersicht aller Mitgliedsorganisationen des Paritätischen Berlin. Einzelne Organisationen finden Sie über die alphabetische Reihenfolge oder über Suchbegriffe.",
    "Notwendige Cookies",
    "Statistik",
    "Externe Inhalte"
]

for offset in range(0, 820, 10):
    url = f"{BASE_URL}?tx_lfmembers_list[offset]={offset}"
    print(f"🔹 Lade Seite mit Offset={offset} …")

    r = requests.get(url)
    if r.status_code != 200:
        print(f"⚠️ Fehler bei {url}: Status {r.status_code}")
        continue

    soup = BeautifulSoup(r.text, "html.parser")
    orgs = soup.find_all("h3")

    if not orgs:
        print(f"⚠️ Keine Organisationen auf Seite {offset} gefunden")
        continue

    for org in orgs:
        name = org.get_text(strip=True)

        # Falls einer der Blacklist-Begriffe vorkommt → überspringen
        if any(bad in name for bad in blacklist):
            continue

        results.append({"Name": name})

    print(f"✅ {len(orgs)} Organisationen (bisher insgesamt {len(results)})")
    sleep(1)

print(f"\n🎉 Fertig! Insgesamt {len(results)} Organisationen gesammelt.")

# Ergebnisse als DataFrame anzeigen
df = pd.DataFrame(results)
display(df.head(20))
print(f"Gesamtzahl der Organisationen: {len(df)}")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from time import sleep

BASE_URL = "https://www.paritaet-berlin.de/mitglieder/unsere-mitgliedsorganisationen"
results = []

blacklist = [
    "Hier finden Sie eine Übersicht aller Mitgliedsorganisationen des Paritätischen Berlin. Einzelne Organisationen finden Sie über die alphabetische Reihenfolge oder über Suchbegriffe.",
    "Notwendige Cookies",
    "Statistik",
    "Externe Inhalte"
]

for offset in range(0, 820, 10):
    url = f"{BASE_URL}?tx_lfmembers_list[offset]={offset}"
    print(f"🔹 Lade Seite mit Offset={offset} …")

    r = requests.get(url)
    if r.status_code != 200:
        print(f"⚠️ Fehler bei {url}: Status {r.status_code}")
        continue

    soup = BeautifulSoup(r.text, "html.parser")

    # Jeder Organisationseintrag steckt in div.members__item
    items = soup.find_all("div", class_="members__item")
    if not items:
        print(f"⚠️ Keine Organisationen auf Seite {offset} gefunden")
        continue

    for item in items:
        name_tag = item.find("h3")
        link_tag = item.find("a")

        if not name_tag or not link_tag:
            continue

        name = name_tag.get_text(strip=True)
        link = link_tag.get("href")

        if any(bad in name for bad in blacklist):
            continue

        if link and not link.startswith("http"):
            link = "https://www.paritaet-berlin.de" + link

        results.append({"Name": name, "Link": link})

    print(f"✅ {len(items)} Organisationen (bisher insgesamt {len(results)})")
    sleep(1)

print(f"\n🎉 Fertig! Insgesamt {len(results)} Organisationen gesammelt.")

df_links = pd.DataFrame(results)
display(df_links.head(20))
print(f"Gesamtzahl der Organisationen: {len(df_links)}")
```
